# LSG workflow plots

Reads `outputs/evaluation/workflow_summary.json` and `pred_examples.npz` produced by `scripts/run_lsg_workflow.py`.

Until licensed TUFLOW cubes are in `data/raw/hf` (and `lf120` / `lf300`), figures are a **synthetic demo**, not Brisbane hydraulics.

```powershell
python scripts/run_lsg_workflow.py --config config/brisbane.yaml --synthetic
python scripts/plot_workflow_results.py
```

````python
from pathlib import Path
import sys

root = Path.cwd()
if not (root / "scripts" / "plot_workflow_results.py").exists():
    root = Path("..").resolve()
sys.path.insert(0, str(root / "scripts"))

from plot_workflow_results import _load_summary, plot_results

summary_path = root / "outputs" / "evaluation" / "workflow_summary.json"
summary = _load_summary(summary_path)
examples = Path(summary.get("pred_examples") or summary_path.parent / "pred_examples.npz")
out_dir = root / "outputs" / "figures"
paths = plot_results(summary, examples, out_dir)
print("data_mode:", summary.get("data_mode"), "gpflow:", summary.get("gpflow_enabled"))
for p in paths:
    print(p)
````
